```python
df = df.dropna()
df
```

```python
df[df["bedrooms"] == 0]
```

```python
df[df["bathrooms"] == 0]
```

```python
df = df[(df["bedrooms"] > 0) & (df["bathrooms"] > 0)]
df
```

```python
df["date"] = pd.to_datetime(df["date"])
df
```

```python
df["Sale Year"] = df["date"].dt.year
df
```

```python
df["Property Age"] = df["Sale Year"] - df["yr_built"]
df
```

```python
df["Is Renovated"] = np.where(df["yr_renovated"] > 0, "Yes", "No")
df
```

```python
df["Price per Square Foot Living"] = df["price"] / df["sqft_living"]
df
```

```python
df["Price per Square Foot Lot"] = df["price"] / df["sqft_lot"]
df
```

```python
df.head()
```

```python
df["price"].agg(["mean", "median", "min", "max"])
```

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["price"] < lower) | (df["price"] > upper)]

outliers
```

```python
outliers.shape
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["price"])

plt.title("House Price Outliers")
plt.xlabel("House Price")
plt.show()
```
